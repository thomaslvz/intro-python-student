
## Un exemple

Une condition est une expression qui vaut `True` ou `False`. Python exécute le bloc indenté après `if` seulement si la condition est vraie.

👉 Exécutez les cellules suivantes. Que constatez-vous lorsque vous changez la valeur de `revenue` ?

```python
revenue = 1200
if revenue > 1000:
    print("Good month")
```

La clause `else` permet de prévoir ce qui se passe lorsque la condition du `if` est fausse. Il y a donc deux "branches" : 
- la branche du `if`
- la branche du `else`

Les deux branches sont exclusives : une seule est exécutée.

👉 Comme précédemment, faites varier la valeur de `revenue` pour comprende la syntaxe `if: ... else: ...`

```python
revenue = 1200
if revenue > 1000:
    print("Good month")
else:
    print("Bad month")
```
