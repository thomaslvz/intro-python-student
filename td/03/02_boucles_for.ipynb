## La syntaxe `range`

Rappel : `range(stop)` produit les entiers de `0` inclus à `stop` exclu. Avec `range(start, stop)`, on commence à `start`, toujours sans inclure `stop`.

👉 Avant d'exécuter les cellules qui suivent, essayez de prédire ce qu'elles vont afficher.

````python
for number in range(5):
    print(number)
````

## Boucler sur une liste avec `enumerate`

`enumerate` fournit à la fois l'indice et la valeur de chaque élément. 

❓ Modifiez la boucle pour produire un dictionnaire qui associe chaque produit à son rang (en commençant à 1). Résultat attendu : `{"Laptop": 1, "Mouse": 2, "Keyboard": 3}`.

````python
products = ["Laptop", "Mouse", "Keyboard"]
ranks = {}

for index, product in enumerate(products):
    print(index)
    print(product)

ranks
````
