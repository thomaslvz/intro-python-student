## Découverte de la syntaxe list comprehension

Observez ce code, et formulez-vous en français à quoi il sert :

````python
sequence = []

for x in [1, 2, 3, 4, 5]:
    sequence.append(x**2)

sequence
````

Dans le code ci-dessus, repérez trois éléments distinctifs, en répondant aux questions suivantes :

1. Où est écrit dans le code l'élément qui est ajouté à chaque tour de boucle ?
2. Quel est l'objet qui est parcouru (on appelle cet objet un itérable) dans toute sa longueur par la boucle ?
3. Comment est noté l'item qui parcourt cet itérable, et qui change de valeur à chaque tour de boucle ?

<details>

<summary>💡 Vérifiez vos réponses : </summary>

1. C'est l'élément `x**2` qui est ajouté à la fin de la liste à chaque tour.
2. L'objet qui est parcouru est la liste `[1, 2, 3, 4, 5]`.
3. L'item qui parcourt cette liste (aussi appelé curseur) est noté `x`
</details>

Observez et exécutez maintenant le code ci-dessous : 

````python
sequence = [x**2 for x in [1, 2, 3, 4, 5]]
sequence
````

La première ligne de code construit la même liste que précédemment, en utilisant la **même boucle**. Seule la syntaxe est différente : elle est plus courte et plus abstraite.

On appelle cette syntaxe une **list comprehension**.

L'image ci-dessous résume le principe de la list comprehension.



<div style='text-align:right'><small><a href="https://datagy.io/list-comprehensions-in-python/" target="_blank">source de l'image</a></small></div>

On peut aussi combiner cette syntaxe avec une condition `if` 🤯.

Regardez le code ci dessous :

````python
sequence = [x**2 for x in [1, 2, 3, 4, 5] if x**2<10]
sequence
````

⬆️ La première ligne de code reproduit de façon très compacte l'intégralité du bloc ci-dessous : 

```python
sequence = []
for x in [1, 2, 3, 4, 5]:
    if x**2 < 10:
        sequence.append(x**2)
```

C'est très pratique pour créer des listes à partir d'un autre objet sur lequel on peut boucler.
